```python
# SECCIÓN 5: Definición de la Clase Completa
class Sala:
    def __init__(self, nombre, capacidad, ubicacion, disponible=True):
        self.nombre = nombre
        self.capacidad = capacidad
        self.ubicacion = ubicacion
        self.disponible = disponible

    def mostrar_info(self):
        estado = "Disponible" if self.disponible else "Reservada"
        print(f"Sala: {self.nombre} | Capacidad: {self.capacidad} | Ubicación: {self.ubicacion} | Estado: {estado}")

    def reservar(self):
        if self.disponible:
            self.disponible = False
            print(f"La sala {self.nombre} ha sido reservada con éxito.")
        else:
            print(f"La sala {self.nombre} ya está ocupada.")

    def liberar(self):
        self.disponible = True
        print(f"La sala {self.nombre} ahora está disponible.")

# SECCIÓN 6: Creación de Objetos y Pruebas

sala1 = Sala("A101", 35, "Edificio A")
sala2 = Sala("Lab Info 2", 25, "Edificio B")
sala3 = Sala("Auditorio Central", 120, "Edificio C")


print("--- Estado Inicial ---")
sala1.mostrar_info()

print("\n--- Intentando Reservar ---")
sala1.reservar()
sala1.mostrar_info()

print("\n--- Intentando Reservar de nuevo (debe dar aviso) ---")
sala1.reservar()
```

## 9. Desafío opcional

Si terminas antes, intenta una o más de estas mejoras:

- agregar el atributo `tipo_sala`
- validar que la capacidad no sea negativa
- mostrar la información en un formato más claro
- crear una lista con varias salas y recorrerla para imprimir sus datos
- impedir que una sala ya reservada se vuelva a reservar sin aviso


```python
# Espacio para desafío opcional
# 9. Desafío opcional: Clase Sala Mejorada
class Sala:
    def __init__(self, nombre, capacidad, ubicacion, tipo_sala="General", disponible=True):
        self.nombre = nombre
        # Validar que la capacidad no sea negativa
        self.capacidad = capacidad if capacidad >= 0 else 0
        self.ubicacion = ubicacion
        self.tipo_sala = tipo_sala
        self.disponible = disponible

    def mostrar_info(self):
        estado = "✅ DISPONIBLE" if self.disponible else "❌ RESERVADA"
        print(f"{'='*40}")
        print(f"SALA: {self.nombre} ({self.tipo_sala})")
        print(f"Ubicación: {self.ubicacion}")
        print(f"Capacidad: {self.capacidad} personas")
        print(f"Estado: {estado}")
        print(f"{'='*40}")

    def reservar(self):
        if self.disponible:
            self.disponible = False
            print(f"📢 Éxito: La sala {self.nombre} ha sido reservada.")
        else:
            # Impedir reserva si ya está ocupada
            print(f"⚠️ Aviso: La sala {self.nombre} YA se encuentra reservada.")

    def liberar(self):
        if not self.disponible:
            self.disponible = True
            print(f"🔓 Éxito: La sala {self.nombre} ha sido liberada.")
        else:
            print(f"ℹ️ La sala {self.nombre} ya estaba disponible.")

# Crear una lista con varias salas y recorrerla
salas_universidad = [
    Sala("A101", 35, "Edificio A", "Sala de clases"),
    Sala("Lab 202", 25, "Edificio B", "Laboratorio"),
    Sala("Magno", 150, "Zócalo", "Auditorio")
]

print("LISTADO COMPLETO DE SALAS:")
for s in salas_universidad:
    s.mostrar_info()

# Prueba de impedimento de reserva doble
print("\nPROBANDO VALIDACIÓN DE RESERVA:")
salas_universidad[0].reservar()
salas_universidad[0].reservar() # Intento de reserva sin aviso previo
```
